# Build the ranked baseline queue

import pandas as pd
import numpy as np
import os

# Use the dataframe created earlier in the notebook
df = df.copy()

# Calculate CTR
df["ctr"] = df["feb_clicks"] / df["feb_impressions"].replace(0, np.nan)
df["ctr"] = df["ctr"].fillna(0)

# Baseline score: high impressions + low CTR = bigger opportunity
df["score"] = (
    np.log1p(df["feb_impressions"]) *
    (1 - df["ctr"].clip(0, 1))
)

# One reason code
df["reason_code"] = np.where(
    df["feb_impressions"] >= df["feb_impressions"].median(),
    "CTR_OPPORTUNITY",
    "LOW_VOLUME"
)

# Action label
df["action"] = np.where(
    df["reason_code"] == "CTR_OPPORTUNITY",
    "REVIEW_CTR",
    "MONITOR"
)

# Rank the queue
queue = df.sort_values("score", ascending=False).reset_index(drop=True)
queue["rank"] = queue.index + 1

# Save the required CSV
os.makedirs("work/outputs", exist_ok=True)
queue.to_csv(
    "work/outputs/baseline_action_score.csv",

)
